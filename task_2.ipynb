```python
import pandas as pd

df = pd.read_csv('orders.csv', encoding='utf-8')

print(df.info())

print("\n\n Первые строки")

print(df.head())
```

```python
def build_status_report(data_frame):
    reformed_data_frame = data_frame.groupby("status").agg(orders=("order_id", "size"), total_amount=("amount", "sum"), avg_amount=("amount", "mean"))
    reformed_data_frame = reformed_data_frame.reset_index()
    return reformed_data_frame.sort_values("total_amount", ascending=False)
```

```python
def load_orders(path):
    df = pd.read_csv(path, encoding='utf-8')
    required_columns = ["order_id", "status", "amount"]
    
    missing_columns = pd.Index(required_columns).difference(df.columns).to_list()
    
    if not missing_columns:
        return df
    
    raise ValueError(f"В датафрейме отсутствуют следующие колонки: {missing_columns}")
```

```python
build_status_report(df).to_csv("status_report.csv", index=False)
```

```python
report_frame = load_orders('orders.csv')


print(report_frame)
```
