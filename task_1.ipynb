```python
def find_max_paid_order(data_frame):
    paid_rows = data_frame[data_frame["status"] == "paid"]
    if paid_rows.empty:
        return None
    max_index = paid_rows["amount"].idxmax()
    return(paid_rows.loc[[max_index], ["order_id", "amount"]])
```

```python
print("Самый дорогой оплаченный заказ:")
print(find_max_paid_order(data_frame))

print("\nКогда оплаченных заказов нет:")
print(find_max_paid_order(data_frame[data_frame["status"] != "paid"]))
```

```python
def summarize_orders(rows):
    amount, orders, paid_orders = 0, len(rows), 0
    for row in rows:
        if row["status"] == "paid":
            amount += row["amount"]
            paid_orders += 1
    
    if paid_orders == 0:
        average_amount = None
    else:
        average_amount = amount / paid_orders
        
    summarize_rows = {"orders": orders, "paid_orders" : paid_orders, "orders_amount" : amount, "average_amount" : average_amount }
    
    return summarize_rows
```

```python
def summarize_by_status(rows):
    
    summarize_rows = {}
    
    for row in rows:
        status = row["status"]
        
        summarize_rows.setdefault(status, {"amount" : 0, "orders" : 0})
        summarize_rows[status]["orders"] += 1
        summarize_rows[status]["amount"] += row["amount"]
        
    return summarize_rows
```
